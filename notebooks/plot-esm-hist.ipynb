```python
import warnings

import matplotlib.pyplot as pl
import numpy as np
import pandas as pd
```

```python
warnings.simplefilter("ignore")
```

```python
models = [
    'ACCESS-ESM1-5',
    'BCC-CSM2-MR',
    'CanESM5-CanOE',
    'CanESM5',
    'CESM2',
    'CNRM-ESM2-1',
    'EC-Earth3-CC',
    'GFDL-ESM4',
    'MIROC-ES2L',
    'MPI-ESM1-2-LR',
    'MRI-ESM2-0',
    'NorESM2-LM',
    'UKESM1-0-LL'
]
```

```python
pl.rcParams['figure.figsize'] = (9/2.54, 8/2.54)
pl.rcParams['font.size'] = 9
pl.rcParams['font.family'] = 'Arial'
pl.rcParams['xtick.direction'] = 'in'
pl.rcParams['ytick.direction'] = 'in'
pl.rcParams['xtick.minor.visible'] = True
pl.rcParams['ytick.minor.visible'] = True
pl.rcParams['ytick.right'] = True
pl.rcParams['ytick.left'] = True
pl.rcParams['xtick.top'] = True
pl.rcParams['axes.spines.right'] = True
pl.rcParams['axes.spines.left'] = True
pl.rcParams['ytick.labelleft'] = True
pl.rcParams['figure.dpi'] = 150
```

```python
results = pd.read_csv('../data/cmip6_esm-hist_co2s.csv', index_col=0)
```

```python
results
```

```python
ghg_cmip6 = pd.read_csv('../data/meinshausen-et-al-2017/Supplementary_Table_UoM_GHGConcentrations-1-1-0_annualmeans_v2.csv', skiprows=21, index_col=0)
ghg_cmip6['CO2']
```

```python
maxco2 = {}
for model in models:
    maxco2[model] = results[model].max()
```

```python
maxco2
```

```python
models_sorted = {k: v for k, v in sorted(maxco2.items(), key=lambda item: item[1], reverse=True)}
models_sorted
```

```python
colors = {model: '#c0c0c0' for model in models}
colors['NorESM2-LM'] = '#D62728'
colors['MPI-ESM1-2-LR'] = '#1973B2'
colors
```

```python
for i, model in enumerate(models_sorted):
    if model not in ['NorESM2-LM', 'MPI-ESM1-2-LR']:
        pl.plot(np.arange(1850.5, 2015), results[model], color=colors[model])

pl.plot(np.arange(1850.5, 2015), ghg_cmip6['CO2'], color='k', label='CMIP6 (Meinshausen et al. 2017)', lw=2)

for model in ['NorESM2-LM', 'MPI-ESM1-2-LR']:
    pl.plot(np.arange(1850.5, 2015), results[model], color=colors[model])

pl.text(1855, 428-0*10, 'MPI-ESM1.2-LR', color=colors['MPI-ESM1-2-LR'])
pl.text(1855, 428-1*10, 'NorESM2-LM', color=colors['NorESM2-LM'])
pl.text(1855, 428-2*10, 'Other CMIP6 models', color='#c0c0c0')

pl.text(1855, 428-4*10, 'CMIP6 best estimate\n(Meinshausen et al. 2017)',  fontweight='bold')

pl.ylabel('ppm')
pl.title('CMIP6 hist-esm CO$_2$ concentrations')

#pl.legend()
pl.xlim(1850, 2015)
pl.ylim(280, 440)
pl.tight_layout()
pl.savefig('../plots/fig4.png')
pl.savefig('../plots/fig4.pdf')
```
